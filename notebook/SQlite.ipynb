```python
import streamlit as st
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("news.db", check_same_thread=False)
c = conn.cursor()
```

```python
c.execute("""
CREATE TABLE IF NOT EXISTS news (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    title TEXT,
    content TEXT,
    label TEXT,
    status TEXT
)
""")
```

```python
conn.commit()
```

```python
c.execute("SELECT * FROM news")
print(c.fetchall())
```

```python
def add_news(title, content, label):
    c.execute(
        "INSERT INTO news (title, content, label, status) VALUES (?, ?, ?, ?)",
        (title, content, label, "pending")
    )
    conn.commit()


def get_news():
    return pd.read_sql_query("SELECT * FROM news", conn)
```

```python
st.title("📰 Fake News Game - Mini Test SQLite")

menu = st.sidebar.selectbox("Menu", ["Ajouter une news", "Voir les news"])
```

```python
if menu == "Ajouter une news":

    st.subheader("➕ Proposer une news")

    title = st.text_input("Titre")
    content = st.text_area("Contenu / lien")
    label = st.selectbox("Type", ["fake", "real", "uncertain"])

    if st.button("Ajouter"):
        add_news(title, content, label)
        st.success("News ajoutée en base ✅")
```
