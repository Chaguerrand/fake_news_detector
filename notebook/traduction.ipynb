```python
from langdetect import detect
from transformers import MarianMTModel, MarianTokenizer

model_name = "Helsinki-NLP/opus-mt-fr-en"

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

def translate_if_needed(text):
    lang = detect(text)

    print("Langue détectée :", lang)

    if lang == "fr":
        inputs = tokenizer(text, return_tensors="pt", padding=True)
        outputs = model.generate(**inputs)
        return tokenizer.decode(outputs[0], skip_special_tokens=True)

    else:
        return text
```

```python
translate_if_needed("Chaque jour, je me battrai pour vous avec chaque souffle de mon corps. Je ne me reposerai pas tant que nous n'aurons pas livré l'Amérique forte, sûre et prospère que nos enfants méritent et que vous méritez. Ce sera véritablement l'âge d'or de l'Amérique.")
```
