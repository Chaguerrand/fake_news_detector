```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.tools.tavily_search import TavilySearchResults

load_dotenv()

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=os.getenv("GOOGLE_API_KEY")
)

search_tool = TavilySearchResults(max_results=5)
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langgraph.prebuilt import create_react_agent

system_prompt = """Tu es un fact-checker expert.

Un modèle ML a déjà classifié l'article. Tu n'as PAS à refaire la classification.

Ton unique mission :
1. Identifie les 2-3 claims principaux de l'article
2. Utilise OBLIGATOIREMENT le tool de recherche pour trouver des sources sur chaque claim
3. Cite OBLIGATOIREMENT les URLs des sources trouvées

Format de réponse finale :
- explication basée sur les sources trouvées
- Sources trouvées :
  - [titre de l'article](URL)
  - [titre de l'article](URL)
  """

fact_checker = create_react_agent(
    llm,
    [search_tool],
    prompt=system_prompt
)
```

```python
result = fact_checker.invoke({
    "messages": [
        ("user", f"""
Classification ML : FAKE

Article : La multiplication par 10 des ventes de véhicules électriques depuis la COP21
        """)
    ]
})
```
